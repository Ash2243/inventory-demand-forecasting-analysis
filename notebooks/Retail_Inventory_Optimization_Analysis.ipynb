### ABC Analysis – Revenue Contribution Segmentation

```python
abc_df = (
    df6.groupby('Description')['SalesDollars']
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

abc_df['cum_pct'] = (
    abc_df['SalesDollars'].cumsum() /
    abc_df['SalesDollars'].sum()
) * 100

abc_df.head(10)
```

```python
def abc_category(x):
    if x <= 80:
        return 'A'
    elif x <= 95:
        return 'B'
    else:
        return 'C'

abc_df['ABC_Category'] = abc_df['cum_pct'].apply(abc_category)

abc_df['ABC_Category'].value_counts()
```

A relatively small subset of products contributes the majority of sales revenue, while a large proportion of SKUs generate comparatively limited financial impact.

```python
abc_df.groupby('ABC_Category')['SalesDollars'].sum().sort_values(ascending=False)
```

A-category products account for the overwhelming majority of sales revenue, reinforcing the importance of prioritizing these SKUs in forecasting, replenishment, and stock availability planning.

## Inventory Turnover Analysis

```python
top_sales = (
    df6.groupby('Description')['SalesQuantity']
    .sum()
    .reset_index()
)

top_inventory = (
    df3.groupby('Description')['onHand']
    .sum()
    .reset_index()
)

turnover_df = top_sales.merge(
    top_inventory,
    on='Description',
    how='inner'
)

turnover_df['inventory_turnover_ratio'] = (
    turnover_df['SalesQuantity'] /
    turnover_df['onHand']
)

turnover_df.sort_values(
    by='inventory_turnover_ratio',
    ascending=False
).head(10)
```

```python
turnover_df = turnover_df[turnover_df['onHand'] > 0]

turnover_df['inventory_turnover_ratio'] = (
    turnover_df['SalesQuantity'] /
    turnover_df['onHand']
)

turnover_df.sort_values(
    by='inventory_turnover_ratio',
    ascending=False
).head(10)
```

```python
turnover_df.sort_values(
    by='inventory_turnover_ratio',
    ascending=True
).head(10)
```

Several products exhibit extremely low inventory turnover despite substantial stock levels, indicating potential overstocking and dead inventory exposure.

```python
df6['SalesDate'] = pd.to_datetime(df6['SalesDate'])

daily_sales = (
    df6.groupby('SalesDate')['SalesQuantity']
    .sum()
    .reset_index()
)

daily_sales.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(14,5))

plt.plot(
    daily_sales['SalesDate'],
    daily_sales['SalesQuantity']
)

plt.title('Daily Sales Quantity Trend')
plt.xlabel('Sales Date')
plt.ylabel('Sales Quantity')

plt.show()
```

Sales activity exhibited a pronounced demand spike followed by gradual normalization, indicating the presence of time-dependent purchasing behavior and potential seasonal or promotional effects.

```python
daily_sales['SalesQuantity'].agg(['mean','std','min','max'])
```

```python
daily_sales['forecast_7day'] = (
    daily_sales['SalesQuantity']
    .rolling(window=7)
    .mean()
    .shift(1)
)

daily_sales[
    ['SalesDate','SalesQuantity','forecast_7day']
].tail(10)
```

The moving-average forecast effectively smooths short-term sales volatility, though it reacts more gradually to sudden demand spikes and drops.

```python
plt.figure(figsize=(14,5))

plt.plot(
    daily_sales['SalesDate'],
    daily_sales['SalesQuantity'],
    label='Actual Sales'
)

plt.plot(
    daily_sales['SalesDate'],
    daily_sales['forecast_7day'],
    label='7-Day Forecast'
)

plt.title('Actual vs Forecasted Sales')
plt.xlabel('Sales Date')
plt.ylabel('Sales Quantity')

plt.legend()

plt.show()
```

```python
from sklearn.metrics import mean_absolute_error
```

```python
forecast_eval = daily_sales.dropna()

mae = mean_absolute_error(
    forecast_eval['SalesQuantity'],
    forecast_eval['forecast_7day']
)

mae
```
